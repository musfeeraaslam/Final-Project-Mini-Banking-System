```python
def check_balance(balance):
  print(f"Check Balance: {balance}")
  return balance

def deposit(balance, amount):
  if amount <= 0:
    print("Deposit Failed!!!!. Amount should be greate than 0")
    return balance

    balance = balance + amount
    print("Deposit Successfull")
    print(f"New Balance: Rs. {balance}")
    return balance

def withdraw(balance, amount):
  if amount <= 0:
    print("Withdrawal failed!!!. Amount should be greate than 0")
    return balance

  if amount > balance:
    print("Withdrawal failed!!!. Amount should be greate than 0")
    return balance

  if balance - amount < 1000:
    print("Withdrawal Failed: Minimum balance of Rs. 1,000 must remain")
    return balance

  balance -= amount
  print("Withdrawal Successful")
  print(f"New Balance: Rs. {balance:.0f}")
  return balance


def calculate_loan_eligibility(age, salary, credit_score):
    if age < 21:
        print("Loan Not Approved: Age must be at least 21")
        return False

    if salary < 50000:
        print("Loan Not Approved: Salary must be at least Rs. 50,000")
        return False

    if credit_score < 650:
        print("Loan Not Approved: Credit score must be at least 650")
        return False

    print("Loan Approved")
    return True


def main():
    balance = 10000
    print(f"Welcome! Your starting balance is Rs. {balance:.0f}")

    while True:
        print()
        print("===== BANK SYSTEM =====")
        print("1. Check Balance")
        print("2. Deposit")
        print("3. Withdraw")
        print("4. Loan Eligibility")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ").strip()

        if choice == "1":
            check_balance(balance)

        elif choice == "2":
            amount = float(input("Enter deposit amount: "))
            balance = deposit(balance, amount)

        elif choice == "3":
            amount = float(input("Enter withdrawal amount: "))
            balance = withdraw(balance, amount)

        elif choice == "4":
            age          = int(input("Enter age: "))
            salary       = float(input("Enter salary: "))
            credit_score = int(input("Enter credit score: "))
            calculate_loan_eligibility(age, salary, credit_score)

        elif choice == "5":
            print("Thank you for using our bank. Goodbye!")
            break

        else:
            print("Invalid choice. Please enter 1-5.")


main()
```
